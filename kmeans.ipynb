```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.datasets import load_iris
import pandas as pd
iris = load_iris()
X = iris.data   # Features (sepal length, sepal width, petal length, petal width)
y = iris.target # Actual labels (for comparison)

feature_names = iris.feature_names

# Convert to DataFrame (optional, better for understanding)
df = pd.DataFrame(X, columns=feature_names)
print(df.head())
```

```python
k = 3
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=k, random_state=42)
kmeans.fit(X)
```

```python
labels = kmeans.labels_
print(labels)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,6))

plt.scatter(X[:,2], X[:,3], c=labels, cmap='viridis', label='Clusters')

plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.title('K-Means Clustering on Iris Dataset')

plt.legend()
plt.show()
```

```python
centroids = kmeans.cluster_centers_

plt.figure(figsize=(8,6))

plt.scatter(X[:,2], X[:,3], c=labels, cmap='viridis')
plt.scatter(centroids[:,2], centroids[:,3],
            s=200, c='red', marker='X', label='Centroids')

plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.title('Clusters with Centroids')

plt.legend()
plt.show()
```

```python
from sklearn.metrics import accuracy_score

# Since labels may be shuffled, we don't directly compare accuracy
# Instead use confusion matrix
from sklearn.metrics import confusion_matrix

print(confusion_matrix(y, labels))
```
