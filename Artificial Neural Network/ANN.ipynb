```python
from tensorflow.keras.datasets import boston_housing
(train_data,train_targets),(test_data,test_targets)=(boston_housing.load_data())
print(train_data.shape)
print(test_data.shape)
```

```python
mean=train_data.mean(axis=0)
train_data-=mean
std=train_data.std(axis=0)
train_data/=std
test_data-=mean
test_data/=std
```

```python
from tensorflow.keras import layers
import keras
model=keras.Sequential([
layers.Dense(64,activation='relu'),  #Dense-fully ConnectionError
layers.Dense(64,activation='relu'),
layers.Dense(1)
])
```

```python
model.compile(optimizer="rmsprop",loss="mse",metrics=["mae"])  #compilation
```

```python
#training
model.fit(train_data,train_targets,epochs=130,batch_size=16,verbose=0)  #16 in 1 row  #epoch-complete training set
```

```python
#evaluate the model
test_mse_score =model.evaluate(test_data,test_targets)
```

```python
results=model.evaluate(test_data,test_targets)
results
```
